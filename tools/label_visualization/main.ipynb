```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
df = pd.read_excel("/run/user/678027/gvfs/sftp:host=access.sophia.grid5000.fr,user=dlai/srv/storage/stars@storage3.sophia.grid5000.fr/dlai/Dataset/DanceClip_TAG/stress_label_84.xlsx")

# Set index
df = df.set_index("Numéro")

## [1]
# Get 'Age' column and clean
ages = df["Age"].dropna()

# Draw histogram
plt.figure(figsize=(7, 5))
plt.hist(ages, bins=range(int(ages.min()), int(ages.max()) + 2), edgecolor='black', alpha=0.7)

# Labels and title
plt.xlabel("Age (years)")
plt.ylabel("Number of dancers")
plt.title("Age Distribution of Dancers")

# Optional grid
plt.grid(axis='y', linestyle='--', alpha=0.6)

plt.show()


## [2]
genders = df["Genre"].dropna()

# Count each gender
gender_counts = genders.value_counts()

# Separate for color mapping
boys = gender_counts.get("M", 0)  # number of males
girls = gender_counts.get("F", 0) # number of females

# Draw bar chart
plt.figure(figsize=(6, 5))
plt.bar("Boy", boys, color='blue', label='Boy')
plt.bar("Girl", girls, color='red', label='Girl')

# Add labels and title
plt.xlabel("Gender")
plt.ylabel("Number of dancers")
plt.title("Gender Distribution of Dancers")
plt.legend()

# Optional grid
plt.grid(axis='y', linestyle='--', alpha=0.6)

plt.show()


## [3]
# Get 'Age' column and clean
ages = df["Stress ressenti plié (EX1)"].dropna()

# Draw histogram
plt.figure(figsize=(7, 5))
plt.hist(ages, bins=range(int(ages.min()), int(ages.max()) + 2), edgecolor='black', alpha=0.7)

# Labels and title
plt.ylabel("Number of dancers")
plt.xlabel("stress level")
plt.title("Distribution of Self-Reported Stress During EX1")

# Optional grid
plt.xlim([0, 10])
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.show()


## [4]
# Get 'Age' column and clean
ages = df["Stress ressenti frappé (EX2)"].dropna()

# Draw histogram
plt.figure(figsize=(7, 5))
plt.hist(ages, bins=range(int(ages.min()), int(ages.max()) + 2), edgecolor='black', alpha=0.7)

# Labels and title
plt.ylabel("Number of dancers")
plt.xlabel("stress level")
plt.title("Distribution of Self-Reported Stress During EX2")

# Optional grid
plt.xlim([0, 10])
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.show()


## [5]
# Get 'Age' column and clean
ages = df["Stress ressenti battement (EX3)"].dropna()

# Draw histogram
plt.figure(figsize=(7, 5))
plt.hist(ages, bins=range(int(ages.min()), int(ages.max()) + 2), edgecolor='black', alpha=0.7)

# Labels and title
plt.ylabel("Number of dancers")
plt.xlabel("stress level")
plt.title("Distribution of Self-Reported Stress During EX3")

# Optional grid
plt.xlim([0, 10])
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.show()


## [6]
# Get 'Age' column and clean
ages = df["stress perçu EX1 juge 1"].dropna()

# Draw histogram
plt.figure(figsize=(7, 5))
plt.hist(ages, bins=range(int(ages.min()), int(ages.max()) + 2), edgecolor='black', alpha=0.7)

# Labels and title
plt.ylabel("Number of dancers")
plt.xlabel("stress level")
plt.title("[PDF] Stress Distribution of Judge1 in EX1")

# Optional grid
plt.xlim([0, 10])
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.show()




## [7]
ages = df["stress perçu EX1 juge 1"].dropna().astype(float)

# Sort values
sorted_ages = np.sort(ages)

# Compute cumulative probabilities
cdf = np.arange(1, len(sorted_ages) + 1) / len(sorted_ages)

# Plot CDF
plt.figure(figsize=(7, 5))
plt.plot(sorted_ages, cdf, linestyle='-', color='blue', linewidth=3)

# Labels and title
plt.xlabel("Stress level")
plt.ylabel("Cumulative probability")
plt.title("[CDF] Stress Distribution of Judge1 in EX1")

# Grid and limits

plt.ylim(0, 1.05)
plt.grid(True, linestyle='--', alpha=0.6)

plt.show()


## [8]
# Get 'Age' column and clean
ages = df["stress perçu EX2 juge 1"].dropna()

# Draw histogram
plt.figure(figsize=(7, 5))
plt.hist(ages, bins=range(int(ages.min()), int(ages.max()) + 2), edgecolor='black', alpha=0.7, color="green")

# Labels and title
plt.ylabel("Number of dancers")
plt.xlabel("stress level")
plt.title("[PDF] Stress Distribution of Judge1 in EX2")

# Optional grid
plt.xlim([0, 10])
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.show()



## [9]
ages = df["stress perçu EX2 juge 1"].dropna().astype(float)

# Sort values
sorted_ages = np.sort(ages)

# Compute cumulative probabilities
cdf = np.arange(1, len(sorted_ages) + 1) / len(sorted_ages)

# Plot CDF
plt.figure(figsize=(7, 5))
plt.plot(sorted_ages, cdf, linestyle='-', color='green', linewidth=3)

# Labels and title
plt.xlabel("Stress level")
plt.ylabel("Cumulative probability")
plt.title("[CDF] Stress Distribution of Judge1 in EX2")

# Grid and limits

plt.ylim(0, 1.05)
plt.grid(True, linestyle='--', alpha=0.6)

plt.show()



## [10]
# Get 'Age' column and clean
ages = df["Exécution EX1 juge 1"].dropna()

# Draw histogram
plt.figure(figsize=(7, 5))
plt.hist(ages, bins=range(int(ages.min()), int(ages.max()) + 2), edgecolor='black', alpha=0.7)

# Labels and title
plt.ylabel("Number of dancers")
plt.xlabel("Execution score")
plt.title("Distribution of Execution Scores by Judge 1 in EX1")

# Optional grid
plt.xlim([0, 20])
plt.xticks(np.arange(int(ages.min())-2, int(ages.max()) + 3, 2))
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.show()



## [11]
# Get 'Age' column and clean
ages = df["Exécution EX2 juge 1"].dropna()

# Draw histogram
plt.figure(figsize=(7, 5))
plt.hist(ages, bins=range(int(ages.min()), int(ages.max()) + 2), edgecolor='black', alpha=0.7, color="green")

# Labels and title
plt.ylabel("Number of dancers")
plt.xlabel("Execution score")
plt.title("Distribution of Execution Scores by Judge 1 in EX2")

# Optional grid
plt.xlim([0, 20])
plt.xticks(np.arange(int(ages.min())-2, int(ages.max()) + 3, 2))
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.show()
```
